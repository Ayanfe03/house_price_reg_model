```python
import pandas as pd
import numpy as np
from sklearn.metrics import mean_absolute_error
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import StandardScaler
```

```python
boston_dataset = pd.read_csv('BostonHousing.csv')
```

```python
boston_dataset.head()
```

```python
boston_dataset.isnull().sum()
```

```python
boston_dataset.describe()
```

```python
boston_dataset.shape
```

```python
X = boston_dataset.drop(columns='MEDV', axis=1)
```

```python
print(X)
```

```python
Y = boston_dataset['MEDV']
```

```python
print(Y)
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, random_state=42)
```

```python
print(X.shape, X_train.shape, X_test.shape)
```

```python
scaler = StandardScaler()
```

```python
scaler.fit(X_train)
X_train_scaled = scaler.transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
print(X_test_scaled)
```

```python
#testing accuracy
linear_reg = LinearRegression()
linear_reg.fit(X_train_scaled, Y_train)
```

```python
X_train_pred = linear_reg.predict(X_train_scaled)
```

```python
X_train_mae = mean_absolute_error(X_train_pred, Y_train)
print('The mean absolute error for training data is:', X_train_mae)
```

```python
X_test_pred = linear_reg.predict(X_test_scaled)
```

```python
X_test_mae = mean_absolute_error(X_test_pred, Y_test)
print('The mean absolute error for test data is:', X_test_mae)
```

```python
# predictive system

input_data = (0.03237,0.0,2.18,0,0.458,6.998,45.8,6.0622,3,222.0,18.7,394.63,2.94)
np_input = np.asarray(input_data)
np_input_reshaped = np_input.reshape(1, -1)
np_scaled = scaler.transform(np_input_reshaped)
print(np_scaled)
```

```python
prediction = linear_reg.predict(np_scaled)
pred_mae = mean_absolute_error(prediction, [33.4])
print(prediction)
print('The mean absolute error of prediction is:', pred_mae)
```
